## Pydantic Basic: Creating and Using Models

Pydantic models are foundation for data validation in Python. They use Python annotations to define the structure and validate data at runtime.

Dataclass is used as a decorator for a class with some attributes.Generally used to assign response of an API using Dataclass

What is the difference between normal class and Data class. A data class does not require a constructor

```python
from dataclasses import dataclass

@dataclass
class Person():
    name: str
    age: int
    city: str
```

```python
person1 = Person(name="Alice", age=30, city="New York")
print(person1)
person2 = Person(name="Bob", age=25, city=35)
print(person2)
```

```python
from pydantic import BaseModel
from typing import Optional

class PersonModel(BaseModel):
    name: str
    age: int
    city: str
```

```python
person3 = PersonModel(name="Alice", age=30, city="New York")
print(person3)
person4 = PersonModel(name="Bob", age=25, city=35)
print(person4)
```

## Model with Optional Fields

- Optional[type] - Indicates the field can be None
- Default value (=None or =True) - Makes the field Optional
- Required Fields must still be provided
- Pydantic validates types even for Optional field when values are provided

```python
class EmployeeModel(BaseModel):
    id: int
    name: str
    department: str
    salary: Optional[float] = None  # Optional field
    is_active: Optional[str] = True  # Optional field with default value
```

```python
emp1 = EmployeeModel(id=1, name="John Doe", department="Engineering")
print(emp1) 
emp2 = EmployeeModel(id=2, name="Jane Smith", department="Marketing", salary="75000")
print(emp2)
```

```python
from typing import List

class ClassroomModel(BaseModel):
    class_name: str
    students: List[str]
    capacity: Optional[int] = None  # Optional field

classroom = ClassroomModel(class_name="Math 101", students=["Alice", "Bob", "Charlie"],capacity="30")
print(classroom)
```

```python
try:
    invalid_classroom = ClassroomModel(class_name="Science 101", students="Alice, Bob, Charlie")
except Exception as e:
    print("Error:", e)
```

## Models with Nested Models

```python
from pydantic import BaseModel

class AddressModel(BaseModel):
    street: str
    city: str
    zip_code: str

class CustomerModel(BaseModel):
    id: int
    name: str
    address: AddressModel  # Nested model
```

```python
customer1 = CustomerModel(id=1, name="Emily Davis", address=AddressModel(street="123 Main St", city="Springfield", zip_code="12345"))
print(customer1)
```

## Pydantic Fields: Customizations and Constraints

```python
from pydantic import BaseModel, Field

class ItemModel(BaseModel):
    name: str = Field(..., title="Item Name", max_length=50)
    price: float = Field(..., gt=0, description="Price must be greater than zero")
    description: Optional[str] = Field(None, max_length=300)    
```

```python
item1 = ItemModel(name="Laptop", price=999.99, description="A high-end gaming laptop.")
print(item1)
```
